```python
import cv2
import numpy as np
from ultralytics.utils.plotting import Annotator, colors

# Define skeleton connections for COCO keypoints
skeleton = [[16, 14], [14, 12], [17, 15], [15, 13], [12, 13], [6, 12], [7, 13], [6, 7], [6, 8], [7, 9], [8, 10], [9, 11], [2, 3], [1, 2], [1, 3], [2, 4], [3, 5], [4, 6], [5, 7]]

# Get the original video dimensions
# Assuming results[0].orig_shape gives (height, width)
if results and results[0].orig_shape:
    height, width = results[0].orig_shape
    fps = 30  # Assuming a standard frame rate, adjust if needed
    output_filename = "skeleton_video.avi"
    fourcc = cv2.VideoWriter_fourcc(*'XVID')
    out = cv2.VideoWriter(output_filename, fourcc, fps, (width, height))

    print(f"Generating skeleton video: {output_filename}")

    for r in results:
        frame = np.zeros((height, width, 3), dtype=np.uint8)  # Create a black frame
        if hasattr(r, 'keypoints') and r.keypoints is not None:
            keypoints = r.keypoints.data[0].cpu().numpy()  # Get keypoints as numpy array

            # Draw keypoints
            for i in range(keypoints.shape[0]):
                x, y, conf = keypoints[i]
                if conf > 0.5:  # Draw keypoint if confidence is above a threshold
                    cv2.circle(frame, (int(x), int(y)), 5, (0, 255, 0), -1) # Green circle

            # Draw skeleton
            for connection in skeleton:
                kpt1_idx, kpt2_idx = connection
                x1, y1, conf1 = keypoints[kpt1_idx - 1] # Adjust index for 0-based array
                x2, y2, conf2 = keypoints[kpt2_idx - 1] # Adjust index for 0-based array
                if conf1 > 0.5 and conf2 > 0.5: # Draw connection if both keypoints are confident
                    cv2.line(frame, (int(x1), int(y1)), (int(x2), int(y2)), (255, 0, 0), 2) # Blue line

        out.write(frame)

    out.release()
    print("Skeleton video generated successfully!")
else:
    print("Could not get video dimensions or no results available.")
```

```python
import cv2
import numpy as np

# Define skeleton connections for COCO keypoints (1-based indexing)
skeleton = [[16, 14], [14, 12], [17, 15], [15, 13], [12, 13], [6, 12], [7, 13], [6, 7], [6, 8], [7, 9], [8, 10], [9, 11], [2, 3], [1, 2], [1, 3], [2, 4], [3, 5], [4, 6], [5, 7]]

# Assuming 'poses' is a list of lists, where poses[0] contains keypoints for person 1 for each frame
if poses and len(poses) > 0 and len(poses[0]) > 0:
    # Get video dimensions from the first frame of the original results if available, otherwise use defaults
    if results and results[0].orig_shape:
        height, width = results[0].orig_shape
    else:
        # Estimate dimensions based on the maximum keypoint coordinates if original dimensions are not available
        max_x = 0
        max_y = 0
        for frame_poses in poses[0]:
            for kp in frame_poses:
                max_x = max(max_x, kp[0])
                max_y = max(max_y, kp[1])
        height, width = int(max_y + 50), int(max_x + 50) # Add some padding

    fps = 30  # Assuming a standard frame rate, adjust if needed
    output_filename = "person1_skeleton_video.avi"
    fourcc = cv2.VideoWriter_fourcc(*'XVID')
    out = cv2.VideoWriter(output_filename, fourcc, fps, (width, height))

    print(f"Generating skeleton video for person 1: {output_filename}")

    for frame_poses in poses[0]:
        frame = np.zeros((height, width, 3), dtype=np.uint8)  # Create a black frame

        # Draw keypoints
        for i in range(frame_poses.shape[0]):
            x, y = frame_poses[i]
            cv2.circle(frame, (int(x), int(y)), 5, (0, 255, 0), -1) # Green circle

        # Draw skeleton
        for connection in skeleton:
            kpt1_idx, kpt2_idx = connection
            # Ensure indices are within the bounds of the keypoints array
            if kpt1_idx - 1 < frame_poses.shape[0] and kpt2_idx - 1 < frame_poses.shape[0]:
                x1, y1 = frame_poses[kpt1_idx - 1] # Adjust index for 0-based array
                x2, y2 = frame_poses[kpt2_idx - 1] # Adjust index for 0-based array
                cv2.line(frame, (int(x1), int(y1)), (int(x2), int(y2)), (255, 0, 0), 2) # Blue line
            else:
                print(f"Warning: Skipping connection {connection} due to out-of-bounds keypoint index.")


        out.write(frame)

    out.release()
    print("Person 1 skeleton video generated successfully!")
else:
    print("No pose data available for person 1.")
```

```python
# --- Pose extraction for all videos in a folder ---
import os, glob, time, csv
import numpy as np
import pandas as pd
from pathlib import Path
import torch
from ultralytics import YOLO
from tqdm import tqdm

# ============ SETTINGS ============
VIDEO_GLOB = "/content/shoplifting-video-dataset/normal/*.mp4"
OUT_DIR     = Path("/content/drive/MyDrive/pose_exports")  # <- saves to Drive
MODEL_PATH  = "yolov8n-pose.pt"    # smallest & fastest
IMG_SIZE    = 640                  # reduce (512/448) for more speed
VID_STRIDE  = 2                    # process every 2nd frame (higher = faster)
MAX_FRAMES_PER_VIDEO = None        # e.g. 200 for testing
# =================================

OUT_DIR.mkdir(parents=True, exist_ok=True)

# Use GPU if available
device = 0 if torch.cuda.is_available() else "cpu"
print("Using device:", "cuda" if device == 0 else "cpu")

# Load model
model = YOLO(MODEL_PATH)
if device == 0:
    model.to("cuda")

# Find videos
videos = sorted(glob.glob(VIDEO_GLOB))
print("Found", len(videos), "videos")

def write_rows_csv(csv_path: Path, rows, header):
    """Append rows to CSV, creating header if needed."""
    is_new = not csv_path.exists()
    with open(csv_path, "a", newline="") as f:
        w = csv.DictWriter(f, fieldnames=header)
        if is_new:
            w.writeheader()
        w.writerows(rows)

def xy_rows_from_result(result, video_name: str, frame_idx: int):
    """Return list of dict rows: video, frame, person, x_0, y_0, ..."""
    out = []
    kp = getattr(result, "keypoints", None)
    if kp is None or kp.data is None:
        return out
    data = kp.data  # [P, K, 3]
    if isinstance(data, torch.Tensor):
        xy = data[:, :, :2].detach().cpu().numpy()
    else:
        xy = np.asarray(data)[:, :, :2]

    if xy.size == 0:
        return out

    num_people, num_kpts = xy.shape[:2]
    for p in range(num_people):
        flat = xy[p].reshape(-1)  # x0,y0,x1,y1,...
        row = {"video": video_name, "frame": frame_idx, "person": p}
        for k in range(num_kpts):
            row[f"x_{k}"] = float(flat[2*k])
            row[f"y_{k}"] = float(flat[2*k+1])
        out.append(row)
    return out

# --- Main loop ---
per_video_csvs = []
for vp in videos:
    vname = Path(vp).stem
    per_csv = OUT_DIR / f"{vname}_pose.csv"
    per_video_csvs.append(per_csv)

    if per_csv.exists():
        print(f"Skipping {vname} (already processed)")
        continue

    print(f"\nProcessing: {vp}")
    rows_buffer, header = [], None
    start = time.time()

    frame_iter = model(
        source=vp,
        stream=True,
        verbose=False,
        imgsz=IMG_SIZE,
        vid_stride=VID_STRIDE,
        device=device,
    )

    for frame_idx, res in enumerate(tqdm(frame_iter, desc=vname)):
        if MAX_FRAMES_PER_VIDEO is not None and frame_idx >= MAX_FRAMES_PER_VIDEO:
            break

        rows = xy_rows_from_result(res, Path(vp).name, frame_idx)
        if rows:
            if header is None:
                header = list(rows[0].keys())
            rows_buffer.extend(rows)

        # flush every ~200 frames
        if frame_idx % 200 == 199 and rows_buffer:
            write_rows_csv(per_csv, rows_buffer, header)
            rows_buffer = []

    if rows_buffer:
        write_rows_csv(per_csv, rows_buffer, header)

    print(f"Saved CSV → {per_csv} | time: {time.time()-start:.1f}s")

# --- Combine all per-video CSVs ---
combined_csv = OUT_DIR / "all_videos_pose_data.csv"
print("\nCombining per-video CSVs into:", combined_csv)
dfs = [pd.read_csv(c) for c in per_video_csvs if c.exists()]
if dfs:
    pd.concat(dfs, ignore_index=True).to_csv(combined_csv, index=False)
    print("Combined rows:", sum(len(d) for d in dfs))
else:
    print("No per-video CSVs found; nothing to combine.")
```

```python
# === Find Shoplifting videos in Google Drive and extract poses (1 cell) ===
import os, re, glob, csv, time
from pathlib import Path
import numpy as np
import pandas as pd
import torch
from ultralytics import YOLO
from tqdm import tqdm
from google.colab import drive

# ---------------- SETTINGS (edit only if you want) ----------------
MODEL_PATH  = "yolov8n-pose.pt"
IMG_SIZE    = 640
VID_STRIDE  = 2                 # higher = faster (every Nth frame)
MAX_FRAMES_PER_VIDEO = None     # e.g., 200 to test; None = full video
LABEL       = "shoplifting"
OUT_BASE    = Path("/content/drive/MyDrive/pose_exports")
# If you already know the exact folder, set it here (uncomment next line):
USER_SHOP_DIR = Path("/content/drive/MyDrive")
# ------------------------------------------------------------------

drive.mount('/content/drive', force_remount=False)

ROOT = Path("/content/drive/MyDrive")

def list_videos(dirpath: Path):
    return sorted([*dirpath.glob("*.mp4"), *dirpath.glob("*.MP4")])

# 1) Choose folder containing shoplifting MP4s
SHOP_DIR = None
try:
    # If user specified a folder, use it
    SHOP_DIR = locals().get("USER_SHOP_DIR", None)
    if SHOP_DIR and list_videos(SHOP_DIR):
        pass
    else:
        SHOP_DIR = None
        # Try to auto-find a folder whose name contains 'shop', with MP4s directly inside
        candidates = []
        for p in ROOT.rglob("*"):
            if p.is_dir() and re.search(r"shop", p.name, re.IGNORECASE) and list_videos(p):
                candidates.append(p)
        if candidates:
            SHOP_DIR = candidates[0]
except Exception:
    SHOP_DIR = None

if SHOP_DIR is None:
    # Fall back: search recursively and show where videos actually are
    all_vids = []
    for pat in ("*.mp4","*.MP4"):
        all_vids += ROOT.rglob(pat)
    if not all_vids:
        raise SystemExit("No MP4 files found anywhere under MyDrive. Check where you placed the dataset.")

    # Show top parent folders that contain videos so you can set USER_SHOP_DIR above
    from collections import Counter
    parents = [v.parent for v in all_vids]
    counts = Counter(parents)
    print("Could not auto-detect a 'Shoplifting' folder. Here are folders that HAVE MP4s:")
    for p, n in counts.most_common(15):
        print(f"{n:4d}  {p}")
    raise SystemExit("\n➡️ Set USER_SHOP_DIR to the correct folder path (copy from above) and rerun this cell.")

print("Using shoplifting folder:", SHOP_DIR)
videos = list_videos(SHOP_DIR)
print("Found", len(videos), "videos")
if not videos:
    raise SystemExit("The chosen folder has 0 MP4s. Open the left file browser to confirm the exact path/casing.")

# 2) Output dir in Drive
OUT_DIR = OUT_BASE / LABEL
OUT_DIR.mkdir(parents=True, exist_ok=True)
combined_csv = OUT_DIR.parent / f"all_{LABEL}_pose_data.csv"

# 3) Device + model
device = 0 if torch.cuda.is_available() else "cpu"
print("Using device:", "cuda" if device == 0 else "cpu")
model = YOLO(MODEL_PATH)
if device == 0:
    model.to("cuda")

def write_rows_csv(csv_path: Path, rows, header):
    is_new = not csv_path.exists()
    with open(csv_path, "a", newline="") as f:
        w = csv.DictWriter(f, fieldnames=header)
        if is_new: w.writeheader()
        w.writerows(rows)

def xy_rows_from_result(result, label: str, video_name: str, frame_idx: int):
    out = []
    kp = getattr(result, "keypoints", None)
    if kp is None or kp.data is None: return out
    data = kp.data
    xy = data[:, :, :2].detach().cpu().numpy() if isinstance(data, torch.Tensor) else np.asarray(data)[:, :, :2]
    if xy.size == 0: return out
    P, K = xy.shape[:2]
    for p in range(P):
        flat = xy[p].reshape(-1)
        row = {"label": label, "video": video_name, "frame": frame_idx, "person": p}
        for k in range(K):
            row[f"x_{k}"] = float(flat[2*k]); row[f"y_{k}"] = float(flat[2*k+1])
        out.append(row)
    return out

# 4) Process videos (resume-safe; per-video CSVs + combined)
per_video_csvs = []
for vp in videos:
    vname = vp.stem
    per_csv = OUT_DIR / f"{vname}_pose.csv"
    per_video_csvs.append(per_csv)
    if per_csv.exists():
        print(f"Skipping {vname} (already processed)")
        continue

    print(f"\nProcessing: {vp}")
    rows_buffer, header = [], None
    start = time.time()

    frame_iter = model(source=str(vp), stream=True, verbose=False, imgsz=IMG_SIZE, vid_stride=VID_STRIDE, device=device)
    for frame_idx, res in enumerate(tqdm(frame_iter, desc=f"{LABEL}:{vname}")):
        if MAX_FRAMES_PER_VIDEO is not None and frame_idx >= MAX_FRAMES_PER_VIDEO:
            break
        rows = xy_rows_from_result(res, LABEL, vp.name, frame_idx)
        if rows:
            if header is None: header = list(rows[0].keys())
            rows_buffer.extend(rows)
        if frame_idx % 200 == 199 and rows_buffer:
            write_rows_csv(per_csv, rows_buffer, header); rows_buffer = []

    if rows_buffer: write_rows_csv(per_csv, rows_buffer, header)
    print(f"Saved CSV → {per_csv} | time: {time.time()-start:.1f}s")

# combine
dfs = [pd.read_csv(c) for c in per_video_csvs if c.exists()]
if dfs:
    pd.concat(dfs, ignore_index=True).to_csv(combined_csv, index=False)
    print("\nCombined CSV:", combined_csv, "| rows:", sum(len(d) for d in dfs))
else:
    print("\nNo per-video CSVs found; nothing to combine.")
```
